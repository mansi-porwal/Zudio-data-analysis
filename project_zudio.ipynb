```python
import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go

```

```python
df = pd.read_csv('/content/Zudio_Synthetic_Sales_10K (1).csv')
df.head()
```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'],dayfirst=True)
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df["Month"]=df['Order_Date'].dt.strftime("%b")
df.head()
#df["Month_Number"]=df['Order_Date'].dt.Month
```

```python
df["Day"]=df["Order_Date"].dt.day_name().str[0:3]
df.sample(5)
```

```python
monthly_sales=(
    df.groupby(['Month_Number',"Month"])['Sales_Amount'].sum().reset_index().sort_values(by="Month_Number")
)
monthly_sales
```

```python

fig=px.line(
    monthly_sales,
    x="Month",
    y="Sales_Amount",
    markers=True,
    template="plotly_dark",
    title="Monthly Sales"
)

fig.update_layout(
    xaxis_title="Month",
    yaxis_title="Sales(Rs.)",
    template="plotly_white"
)

fig.show()

```

```python
fig=px.area(
     monthly_sales,
    x="Month",
    y="Sales_Amount",
    markers=True,
    template="plotly_dark",
    title="Monthly Sales"
)

fig.update_layout(
    xaxis_title="Month",
    yaxis_title="Sales(Rs.)",
    template="plotly_white"
)

fig.show()



```

```python
category_sales=(
    df.groupby("Product_Category")["Sales_Amount"].sum().reset_index()
     .sort_values(by="Sales_Amount",ascending=False)
)
category_sales
```

```python

fig=px.bar(
    category_sales,
    x="Product_Category",
    y="Sales_Amount",
    title="Category Sales",
    template="plotly_white",
    text_auto='.2s',
)

fig.update_layout(
    xaxis_title="Category",
    yaxis_title="Sales(Rs.)",
    template="plotly_white"
)

fig.show()
```

```python
city_sales=(
    df.groupby("City")["Sales_Amount"].sum().reset_index().nlargest(5,"Sales_Amount")
     .sort_values(by="Sales_Amount",ascending=False)
)
city_sales
```

```python
fig = px.pie(
    city_sales,
    names="City",
    values="Sales_Amount",
    title="Top 5 cities by Sales",
    hole=0.5,
    template="plotly_white"
)
fig.update_traces(textinfo="percent+label")
fig.update_layout(title_x=0.5)

fig.show()
```

```python
top_stores = (
    df.groupby("Store",
               as_index=False)["Sales_Amount"].sum().nlargest(5,"Sales_Amount")
               .sort_values(by="Sales_Amount",ascending=False)
)
```

```python
fig = px.bar(
    top_stores,
    x = "Store",
    y = "Sales_Amount",
    title = "Top 5 Stores by Sales",
    template = "plotly_white",
    text = "Sales_Amount"
)
fig.update_layout(
    xaxis_title="Store",
    yaxis_title="Sales(Rs.)",
    title_x=0.5
    )

fig.show()
```

```python
heatmap_data=(
    df.groupby(['Day','Hour'])["Sales_Amount"].sum().reset_index()
)
heatmap_pivot=heatmap_data.pivot(
    index="Day",
    columns="Hour",
    values="Sales_Amount"
)

heatmap_pivot
```

```python
day_order=[
    "Mon","Tue","Wed","Thu","Fri","Sat","Sun"
]

heatmap_pivot=heatmap_pivot.reindex(day_order)
heatmap_pivot
```

```python
fig= go.Figure(
    data = go.Heatmap(
        z=heatmap_pivot.values,
        x=heatmap_pivot.columns,
        y=heatmap_pivot.index,
        colorscale="Viridis"
    )
)

fig.update_layout(
    title="Sales by Day and Hour",
    xaxis_title="Hour",
    yaxis_title="Day"
)
fig.show()
```
